## Visualisation and Loading of Generalisation plots

This notebook loads and visualises the results of the nested cross-validation procedure, including all modelling strategies.

To load the data, please download the `result_sum` folder from the Zenodo record of this paper and place it inside the results directory, as `results/result_sum/`.

```python

import matplotlib.pyplot as plt
import os
import seaborn

from powerlaw_utils import (
    errors_data_outer_fold_generic 
)
```

```python
#General configurations
STATION_NAME_MAP = {
    "outer_station_5": "Collelongo (Italy)",
    "outer_station_12": "Hyytiälä (Finland)",
    "outer_station_14": "Le Bray (France)",
    "outer_station_21": "Sro (Denmark)",
}
#directories
current_dir = os.getcwd()
result_dir= "results/result_sum" #current folder where the results are stored
save_dir = None #os.path.join(current_dir, "test_save") #folder to save visualisation, it set to None, the visualisation will not be saved

#model and error configurations
error_type = 'test_mse' #error type to visualise
pred_types = ['naive', 'regularized', 'residual','bias', 'parallel', 'Mixture', 'process'] #avaible modelling strategies
#number of models per strategy
top_k_configs = 5

#coloring
base_colors = plt.rcParams['axes.prop_cycle'].by_key()['color']
color_map = {pred: base_colors[idx % len(base_colors)] for idx, pred in enumerate(pred_types)}
```

### In distribution (ID) results of Hyytiälä and Collelongo

```python
#ID configurations
stations=[12,5]
station_training_years = {
        5: [1, 2, 3, 5],
        12: [1, 2, 3],
    }
sparse = [50, 30, 7]
sparse_year = 1

mode='temporal'
```

#### Generalisation plots and violin plots

```python
#generate results for ID settings
from Visualisation_generalisation_error_ID import generate_station_plots, plot_coefficent_of_variation

for station in stations:
        training_years = station_training_years.get(station, [1, 2, 3])
        
        #get results for the station and training years
        errors, size = errors_data_outer_fold_generic(
        base_path=result_dir,
        mode=mode,
        lengths=training_years,
        station=station,
        sparse=sparse,
        sparse_year=sparse_year
        )

        #plot gerneralisation error for the station

        result = generate_station_plots(
            station=station,
            training_years=training_years,
            pred_types=pred_types,
            error_type=error_type,
            sparse=sparse,
            sparse_year=sparse_year,
            color_map=color_map,
            top_k_configs=top_k_configs,
            base_path=result_dir,
        )

        plot_coefficent_of_variation(errors,
                                     pred_types, 
                                     error_type, 
                                     color_map, 
                                     station=station,
                                     save_dir=save_dir)

```

### Out of distribution (OOD) results

```python
#generate results for OOD settings

from Visualsation_generalisation_error_OOD import generate_spatial_plots

#OOD configurations
lengths = (3, 6, 9, 12)
sparse = [50, 30,7]
sparse_years = 3
result_dir_ood = os.path.join(current_dir, "results", "result_sum")

generate_spatial_plots(
            lengths=lengths,
            pred_types=pred_types,
            error_type=error_type,
            base_path=result_dir_ood,
            top_k_configs=top_k_configs,
            sparse=sparse,
            sparse_years=sparse_years,
            save_dir=save_dir,
        )

```
